# Neurona de McCulloch-Pitts: Operación OR

Explicación:

La operación OR funciona de manera que la neurona se activa cuando al menos una de las entradas es 1. Si las dos entradas son 0, la neurona no se activa.

En este ejemplo, la neurona recibe dos entradas (x1 y x2). Primero suma las entradas y después compara el resultado con un umbral. Si la suma es mayor o igual al umbral, la salida será 1; de lo contrario, será 0.

La neurona está representando una operación lógica OR porque solamente necesita que una de las entradas sea verdadera para producir una salida de 1.

```python
# Neurona de McCulloch-Pitts para la operación OR

def neurona_or(x1, x2):
    # Pesos de las entradas
    w1 = 1
    w2 = 1

    # Umbral
    umbral = 1

    # Suma ponderada
    suma = (x1 * w1) + (x2 * w2)

    # Función de activación
    if suma >= umbral:
        return 1
    else:
        return 0


# Tabla de verdad de OR
print("x1  x2  Salida")

print(0, " ", 0, " ", neurona_or(0, 0))
print(0, " ", 1, " ", neurona_or(0, 1))
print(1, " ", 0, " ", neurona_or(1, 0))
print(1, " ", 1, " ", neurona_or(1, 1))
```

Conclusión:

La neurona de McCulloch-Pitts puede representar la operación OR utilizando entradas, pesos y un umbral. Cuando por lo menos una entrada tiene el valor 1, la suma alcanza el umbral y la neurona produce una salida de 1.

# Neurona de McCulloch-Pitts: Operación NOT

Explicación:

La operación NOT funciona de manera diferente a OR porque solamente utiliza una entrada y cambia su valor. Si la entrada es 1, la salida será 0. Si la entrada es 0, la salida será 1.

En este ejemplo, la neurona utiliza un peso negativo y un umbral para invertir la entrada. De esta manera podemos representar el comportamiento de una operación NOT mediante una neurona de McCulloch-Pitts.

```python
# Neurona de McCulloch-Pitts para la operación NOT

def neurona_not(x):
    # Peso negativo
    w = -1

    # Umbral
    umbral = 0

    # Suma ponderada
    suma = x * w

    # Función de activación
    if suma >= umbral:
        return 1
    else:
        return 0


# Tabla de verdad de NOT
print("Entrada  Salida")

print(0, "       ", neurona_not(0))
print(1, "       ", neurona_not(1))
```

Conclusión:

La neurona de McCulloch-Pitts puede representar la operación NOT utilizando una sola entrada y un peso negativo. Cuando la entrada es 0, la neurona produce 1, mientras que cuando la entrada es 1, produce 0. Por eso la salida es la inversa de la entrada.
